# EXERCISE 2: Perform a ping measurement to target IP 1

### Use the list of VPs to make measurements towards two targets

```python
VP_IDS = [7193, 6830]
TARGET1 = "142.250.201.36"
TARGET2 = "104.16.124.96"
```

```python
def get_ping_config(
    target: str, vp_ids: list[int], description: str = "GaN ping"
) -> dict:
    """function to define a ping measurement before posting it"""
    return {
        "definitions": [
            {
                "target": target,
                "af": 4,
                "packets": 3,
                "size": 48,
                "tags": ["dioptra", "gan"],
                "description": f"{description}",
                "resolve_on_probe": False,
                "skip_dns_check": True,
                "include_probe_id": False,
                "type": "ping",
            }
        ],
        "probes": [
            {"value": v_id, "type": "probes", "requested": 1} for v_id in vp_ids
        ],
        "is_oneoff": True,
    }
```

## Perform measurement towards the first IP address

```python
url = f"{RIPE_ATLAS_API_URL}/measurements/"

# This time we use the POST method (instead of the GET), to POST a measurement
# on RIPE Atlas using the API
measurement_id_1 = requests.post(
    url=url, headers=HEADER, json=get_ping_config(TARGET1, VP_IDS)
)

print(measurement_id_1.json())
```

#  EXERCISE 4: Perform a traceroute measurement

```python
def get_traceroute_config(
    target: str, vp_ids: list[int], description: str = "GaN ping"
) -> dict:
    """function to define a ping measurement before posting it"""
    return {
        "definitions": [
            {
                "target": target,
                "af": 4,
                "packets": 3,
                "first_hop": 1,
                "max_hops": 30,
                "protocol": "ICMP",
                "tags": ["dioptra", "gan"],
                "description": f"{description}",
                "resolve_on_probe": False,
                "skip_dns_check": True,
                "include_probe_id": False,
                "type": "traceroute",
            }
        ],
        "probes": [
            {"value": v_id, "type": "probes", "requested": 1} for v_id in vp_ids
        ],
        "is_oneoff": True,
    }
```

## Perform traceroute to the first IP address

```python
url = f"{RIPE_ATLAS_API_URL}/measurements/"

# This time we use the POST method (instead of the GET), to POST a measurement
# on RIPE Atlas using the API
measurement_traceroute_1 = requests.post(
    url=url, headers=HEADER, json=get_traceroute_config(TARGET1, VP_IDS)
)

print(measurement_traceroute_1.json())
```

```python
traceroute_id= "212577002"
url = f"{RIPE_ATLAS_API_URL}/measurements/{traceroute_id}/results"

print(url)

# EXERCISE: get results from RIPE Atlas and extract/print the min_rtt for each VP
resp = requests.get(url)
for p in resp.json():
    results = p['result']
    for r in results:
        new_results = []
        liste = r['result']
        copie = liste
        print("nouvelle copie : ",copie)
        while copie != []:
            min=copie[0]
            if copie!= [{'x': '*'}, {'x': '*'}, {'x': '*'}]:
                min_rtt = min['rtt']
                for courant in copie:
                    #print("courant_rtt : ",courant['rtt'])
                    if min_rtt > courant['rtt']:
                        min_rtt = courant['rtt']
                        min=courant
                        print("min_rtt : ",min_rtt)
                new_results.append(min)
                copie.remove(min)
        print("new_results : ",new_results)
    
    print("results : ",new_results,", min : ",p['min'])
```

## Retrieve the results and print the IP level path for each VP

```python
for p in resp.json():
    results = p['result']
    print("path of ",p['prb_id']," : ")
    for r in results :
        froms = r['result']
        for fr in froms:
            if fr.get('from')!= None:
                print(" --> ",fr.get('from'),end="")
    print("")
```
