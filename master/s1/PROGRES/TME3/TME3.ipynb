## Calendar convention

- Use regular expressions to transform a date from MM-DD-YYYY format to DD-MM-YYYY 
format. (example "11-06-2020" should become "06-11-2020").
- (*) Do the same thing using the `datetime` package.

Answer

```python
def switch_md(txt):
    return re.sub(r'(.*)-(.*)-(.*)', r'\2-\1-\3', txt)
```

```python
switch_md("11-06-2020")
```

# Exercice 2 - Analyze XML

- Write a Python code that retrieves the content of the page at:

```python
url = "https://www.w3schools.com/xml/cd_catalog.xml"
```

- Look at the text content and load as xml.

Answer

```python
from requests import Session
from bs4 import BeautifulSoup 
s = Session()
r = s.get(url)
r.encoding = "utf8"
catalog = r.text
soup = BeautifulSoup(catalog, 'xml')

catalog = soup.find_all('CD')
print(catalog)
```

- Write a `display_cd` function that displays (i.e. `print`), for a CD: title, artist, country, company, year.
- Display all CDs.

Answer

```python
def display_cd(entry):
    titles = entry.find_all('TITLE')
    artists = entry.find_all('ARTIST')
    countries = entry.find_all('COUNTRY')
    companies = entry.find_all('COMPANY')
    years = entry.find_all('YEAR')
    return titles[0].text +", by " + artists[0].text +" from " + countries[0].text +" ("+ years[0].text +"). (c) " + companies[0].text
```

```python
for cd in catalog:
    print(display_cd(cd))
```

- Display all 1980s CDs. 

Answer

```python
for cd in catalog:
    disque = display_cd(cd)
    if re.fullmatch(r'(.*)(\(198\d\))(.*)', disque) is not None:
        print(disque)
```

- Display all British CDs.

Answer

```python
catalog = soup.find_all('CD')
for cd in catalog:
    if (cd.COUNTRY.text) == "UK":
        print(display_cd(cd))
```

- Write a function that converts an entry in a string that shows director, title, district, start date, end date, and geographic coordinates.
- Convert all entries in strings (warning: some entries may have issues).
- Display the first 20 entries.

Answer

```python
def display_loc(entry):
    fields = entry.get("fields")
    director = fields.get("nom_realisateur")
    title = fields.get("nom_tournage")
    district = fields.get("ardt_lieu")
    start_date = fields.get("date_debut")
    end_date = fields.get("date_fin")
    geo_shape = fields.get("geo_shape")
    coordinates = geo_shape.get("coordinates") 

    director = director if director is not None else "N/A"
    title = title if title is not None else "N/A"
    district = district if district is not None else "N/A"
    start_date = start_date if start_date is not None else "N/A"
    end_date = end_date if end_date is not None else "N/A"
    
    return f"{title}, by {director}, in district {district}, from {start_date} to {end_date}, coordinates: {coordinates}"
```

```python
all_entries = [display_loc(e) for e in locs]
print('\n'.join(all_entries[:20]))
```

```python
movies = dict()

for entry in locs:
    fields = entry.get("fields")
    
    title = fields.get("nom_tournage")
    director = fields.get("nom_realisateur")
    district = fields.get("ardt_lieu")
    start_date = fields.get("date_debut")
    end_date = fields.get("date_fin")
    
    clean = title.strip().lower()
    
    if clean not in movies:
        movies[clean] = { "title": title,"director": director,"locations": []}
    
    movies[clean]["locations"].append({"district": district, "start_date": start_date, "end_date": end_date})

movies = [m for m in movies.values()]
```

```python
len(movies)
```

```python
def display_movie(movie):
    title = movie.get("title")
    director = movie.get("director")

    locations = movie.get("locations")
    nb_shootings = len(locations)
    start_dates = [loc.get("start_date") for loc in locations]
    end_dates = [loc.get("end_date") for loc in locations]
        
    earliest_start = min(start_dates)
    latest_end = max(end_dates)
    
    return f'Directeur : {director}, Titre : "{title}", Nombre de Shootings : {nb_shootings} from {earliest_start} to {latest_end}'
```

```python
all_movie_displays = [display_movie(m) for m in movies]
print('\n'.join(all_movie_displays[:20]))
```

```python
stats = dict()

for entry in locs:
    fields = entry.get("fields")
    
    district = fields.get("ardt_lieu")
    
    if district not in stats:
        stats[district] = 0
    
    stats[district] += 1

stats = dict(sorted(stats.items(), key=lambda item: item[1], reverse=True)) # trier par nombre de tournages decroissant (ligne ajoutée par l'ia intégrée dans Visual Studio Code sans prompt)

for d, n in stats.items():
    print(f"Number of shootings in {d}: {n}.")
```
